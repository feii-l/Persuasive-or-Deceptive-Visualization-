```python
import pandas as pd
import plotly.express as px
```

```python
df= pd.read_csv("abortion_clean.csv")

df.head()
```

```python
fig = px.choropleth(df, locations="state_abbr", locationmode="USA-states", scope="usa",
                    color="pct_counties_no_clinic_2020", color_continuous_scale="Reds", range_color=(0,100),
                    hover_name="state",
                    labels={"pct_counties_no_clinic_2020": "% of counties<br>with no clinic"},
                    title="In the average state, 78% of counties have no abortion clinic (2020)")
fig.update_geos(showlakes=False)
fig.update_coloraxes(colorbar_ticksuffix="%")
fig.update_traces(hovertemplate="<b>%{hovertext}</b><br>%{z}% of counties have no clinic<extra></extra>")
fig.write_image("pro_plot1_map.png", scale=2)
fig
```

```python
# only the 17 states that lost clinics (cherry-picked)
lost = df[df.pct_chg_clinics_2017_2020 < 0]
clinics_2020 = lost.clinics_2020.sum()
clinics_2017 = (lost.clinics_2020 - lost.chg_clinics_2017_2020).sum()

totals = pd.DataFrame({"year": ["2017", "2020"], "clinics": [clinics_2017, clinics_2020]})
fig2 = px.bar(totals, x="year", y="clinics", text="clinics",
              color_discrete_sequence=["#b2182b"],
              labels={"year": "", "clinics": "Abortion clinics"},
              title="Abortion clinics are disappearing across 17 states")
fig2.update_traces(texttemplate="%{y:.0f}", textposition="outside")
fig2.update_yaxes(range=[340, 405])  # truncated axis: an 11% drop looks like a collapse
fig2.update_layout(width=600, height=500)
fig2.write_image("pro_plot2_clinics.png", scale=2)
fig2
```
